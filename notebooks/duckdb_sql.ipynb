````python
import duckdb
````

````python
conn = duckdb.connect("../data/credit_risk.db")
````

## Database Structure

```text
credit_risk.duckdb
│
├── raw
│   └── loans
│
├── staging
│   └── loans_clean
│
└── analytics
    ├── pd_model_data
    ├── expected_loss
    ├── portfolio_summary
    └── monitoring
```

- **raw** → Original source data
- **staging** → Cleaned and standardized data
- **analytics** → Model-ready and analytical datasets

````python
# Create the schema

conn.execute("""
CREATE SCHEMA IF NOT EXISTS raw;
CREATE SCHEMA IF NOT EXISTS staging;
CREATE SCHEMA IF NOT EXISTS analytics;
""")
````

````python
# Create the raw loans table from the original Lending Club CSV
conn.execute("""
CREATE OR REPLACE TABLE raw.loans AS
SELECT *
FROM read_csv(
    '../data/loan_data_2007_2014.csv',
    sample_size = -1
);
""");
````
