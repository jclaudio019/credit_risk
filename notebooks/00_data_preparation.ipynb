# 00. Data Preparation

This first notebook loads the Lending Club file and checks that the fields used in the later PD workflow are present.

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('../data/loan_data_2007_2014.csv')
loan_data = pd.read_csv(DATA_PATH, low_memory=False)
print(f'Loaded {DATA_PATH}: {loan_data.shape[0]:,} rows x {loan_data.shape[1]:,} columns')
loan_data.dtypes.head(15)
```

```python
required_columns = ['loan_status', 'emp_length', 'term', 'earliest_cr_line', 'issue_d',
                    'annual_inc', 'funded_amnt', 'total_rev_hi_lim', 'grade', 'int_rate', 'dti']
missing_columns = sorted(set(required_columns) - set(loan_data.columns))
if missing_columns:
    raise ValueError(f'Missing required columns: {missing_columns}')
print('Required columns are present.')
loan_data[required_columns].head()
```
