# Data Understanding and Preparation

This notebook examines Lending Club loans issued from 2007 through 2014. It establishes the raw-data context and prepares reproducible representations of employment length, loan term, and credit-history dates for the PD analysis.

```python
from pathlib import Path

import numpy as np
import pandas as pd

pd.options.display.max_columns = None
REFERENCE_DATE = pd.Timestamp("2017-12-01")
DATA_PATH = Path("../data/loan_data_2007_2014.csv")
loan_data = pd.read_csv(DATA_PATH, low_memory=False)
```

## Dataset overview

The dataset contains historical consumer-loan records. The fields below provide the initial view of dimensions, observations, and available variables before cleaning.

```python
print(f"Rows: {loan_data.shape[0]:,}; columns: {loan_data.shape[1]:,}")
display(loan_data.head())
display(loan_data.tail())
loan_data.columns.to_list()
```

```python
loan_data.info()
```

## Missingness overview

Missingness is reviewed before any repair. The next notebook makes a separate decision for each field type: a same-row proxy for revolving credit limit, a training-only statistic for income, zero for selected event counts, and retained missingness where absence can be analytically informative. This separation avoids treating all missing values as the same problem.

```python
missingness = (
    loan_data.isna().sum()
    .to_frame("missing_count")
    .assign(missing_rate=lambda frame: frame["missing_count"] / len(loan_data))
    .sort_values("missing_count", ascending=False)
)
display(missingness[missingness["missing_count"] > 0])
```

## Employment length and loan term

Employment length and term arrive as strings. Numeric extraction makes both fields usable in later analysis. The values `10+ years` and `< 1 year` yield their numeric components; missing employment length remains missing at this stage and is addressed only after the train/test split. Loan term contains the expected 36- and 60-month values.

```python
display(loan_data["emp_length"].value_counts(dropna=False).sort_index())
loan_data["emp_length_int"] = loan_data["emp_length"].str.extract(r"(\d+)", expand=False).astype("float")

display(loan_data["term"].value_counts(dropna=False).sort_index())
loan_data["term_int"] = loan_data["term"].str.extract(r"(\d+)", expand=False).astype("int")
loan_data[["emp_length", "emp_length_int", "term", "term_int"]].head()
```

## Credit-history and issue dates

The source dates use two-digit years. Dates later than the fixed analysis reference date are interpreted as historical dates and moved back one century before months-since values are calculated. This replaces the original practice of assigning those observations the maximum observed duration.

```python
loan_data["earliest_cr_line_date"] = pd.to_datetime(
    loan_data["earliest_cr_line"], format="%b-%y", errors="coerce"
)
pre_correction_months = np.round(
    (REFERENCE_DATE - loan_data["earliest_cr_line_date"])
    / np.timedelta64(1, "D")
    / 30.4375
)
future_credit_line = loan_data["earliest_cr_line_date"] > REFERENCE_DATE
affected_credit_line = loan_data.loc[
    future_credit_line, ["earliest_cr_line", "earliest_cr_line_date"]
].assign(mths_earliest_cr_line_date=pre_correction_months[future_credit_line])

print("Pre-correction months-since summary")
display(pre_correction_months.describe().to_frame("mths_earliest_cr_line_date"))
print(f"Affected future-parsed dates: {future_credit_line.sum():,}")
display(affected_credit_line.head())

loan_data.loc[future_credit_line, "earliest_cr_line_date"] = (
    loan_data.loc[future_credit_line, "earliest_cr_line_date"] - pd.DateOffset(years=100)
)
loan_data["mths_earliest_cr_line_date"] = np.round(
    (REFERENCE_DATE - loan_data["earliest_cr_line_date"])
    / np.timedelta64(1, "D")
    / 30.4375
)
print("Post-correction months-since summary")
display(
    loan_data["mths_earliest_cr_line_date"]
    .describe()
    .to_frame("mths_earliest_cr_line_date")
)

loan_data["issue_date"] = pd.to_datetime(
    loan_data["issue_d"], format="%b-%y", errors="coerce"
)
loan_data["mths_issue_date"] = np.round(
    (REFERENCE_DATE - loan_data["issue_date"])
    / np.timedelta64(1, "D")
    / 30.4375
)
display(
    loan_data[
        [
            "earliest_cr_line",
            "earliest_cr_line_date",
            "mths_earliest_cr_line_date",
            "issue_d",
            "issue_date",
            "mths_issue_date",
        ]
    ].head()
)
```

## Conclusions

The raw data contain the core loan, borrower, credit-history, and outcome fields needed for the historical PD exercise. Employment length, term, and date fields now have reproducible numeric or datetime representations. The future-date correction retains each borrower's actual historical credit duration instead of artificially placing affected rows in the longest-history group.

Missing-value repair and the historical good/bad target are deliberately handled in the next notebook, after the modeling split is defined. That ordering is necessary because an imputation statistic estimated from all rows would allow information from the held-out partition to influence training preparation.
